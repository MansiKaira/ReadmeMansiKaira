```python
df["Gender"] = df["Gender"].replace({"Male": 0, "Female": 1})
df
```

```python
df["Education"] = df["Education"].replace({"High School": 0, "Bachelor": 1, "Master": 2, "PhD": 3})
df
```

```python
df['Location'] = df['Location'].replace({"Rural": 0, "Suburban": 1, "Urban": 2})
df
```

```python
df['Job_Title'] = df['Job_Title'].replace({"Analyst": 0, "Manager": 1, "Engineer": 2, "Director": 3})
df
```

```python
print(df.columns.tolist())
```

```python
X = df[['Education','Job_Title','Experience']]
```

```python
X
```

```python
y = df['Salary']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)
```

```python
X_train
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
numeric_cols = ['Experience']
```

```python
X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
```

```python
X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred[:5]
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
```

```python
df['Salary'].mean()
```

```python
sns.histplot(data=df, x='Salary', bins=20, kde=True)
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
RMSE = np.sqrt(mean_squared_error(y_test, y_pred))
```

```python
RMSE
```

```python
y_pred = pd.Series(y_pred, index=y_test.index)
test_residuals = y_test - y_pred
```

```python
test_residuals
```

```python
sns.scatterplot(x=y_test, y=test_residuals)
plt.axhline(y=0, color='red', ls='--')
```

```python
sns.displot(test_residuals, bins=25,kde=True)
```

```python
sns.boxplot(x=df.loc[y_test.index, 'Job_Title'], y=test_residuals)
plt.title("Residuals by JobTitle")
plt.show()
```

```python
plt.scatter(y_pred, test_residuals)
```

```python
import scipy as sp
fig, ax = plt.subplots(figsize=(6,8), dpi=100)
_ = sp.stats.probplot(test_residuals, plot=ax)
```
